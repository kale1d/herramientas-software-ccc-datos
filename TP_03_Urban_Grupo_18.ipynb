## Desarrollo del Proyecto Híbrido: Persistencia Relacional y Vectorial

### 1. Reestructuración del Modelo Lógico (Ejercicio 03)
A diferencia del enfoque monolítico que hicimos en un primer enfoque, donde todos los datos del CSV se
almacenaban de forma plana en una única estructura, el nuevo diseño del sistema
implementa el **Principio de Responsabilidad Única (SRP)** y la normalización
de dominios. El modelo se modularizó en cuatro entidades independientes
mediante `@dataclass`:

* **Vehiculo:** Entidad inmutable (`frozen=True`) encargada de aislar la patente.
* **Radar:** Clase especializada que encapsula el ID y la ubicación física del
  dispositivo de captura.
* **Evidencia:** Componente opcional (`Optional`) que agrupa exclusivamente
  los metadatos de la captura visual (`archivo_imagen`, `patente_imagen`, `ratio`).
* **Multa:** Entidad transaccional central que interconecta las estructuras
  anteriores de forma jerárquica mediante referencias de objetos y tipado
  estricto (`datetime.date` y `datetime.time`).


````python
@dataclass(frozen=True)
class Vehiculo:
  """Entidad Vehiculo basada en la patente."""
  patente: str

@dataclass(frozen=True)
class Radar:
  """Entidad Radar identificada por ID y ubicación."""
  radar_id: str
  ubicacion: str

@dataclass
class Evidencia:
  """Entidad Evidencia asociada a una multa de forma opcional."""
  multa_id: str
  archivo_imagen: str
  patente_imagen: Optional[str]
  ratio: Optional[float]

@dataclass
class Multa:
  """Entidad principal que registra la infracción."""
  multa_id: str
  vehiculo: Vehiculo
  radar: Optional[Radar]
  fecha: Optional[date]
  hora: Optional[time]
  ubicacion: str
  velocidad_registrada: float
  velocidad_maxima: float
  exceso_velocidad_real: float
  exceso_velocidad: float
  estado_multa: str
  evidencia: Optional[Evidencia] = None
````

### 2. Parseo y Transformación de Datos (Ejercicio 04)
La función `procesar_fila_csv` fue adaptada para digerir la fila cruda del
archivo y orquestar la instanciación de los sub-objetos del modelo. Se corrigieron
los conflictos de tipos (como la conversión estricta de `multa_id` a entero)
y se implementó lógica defensiva para la conversión segura de valores nulos o
vacíos (`NaN`) provenientes del CSV hacia tipos nativos de Python.

````python
#@title Funciones Auxiliares de Parseo Corregidas (Ejercicio 04)

def is_empty(valor) -> bool:
  """Devuelve True si el valor está vacío, es NaN o un string en blanco."""
  if valor is None:
    return True
  # Control defensivo: si es string y está vacío o lleno de espacios
  if isinstance(valor, str) and valor.strip() == "":
    return True
  try:
    return bool(pd.isna(valor))
  except (TypeError, ValueError):
    return False


def has_value(valor):
  """Convierte strings vacíos, NaN o None en un None unificado."""
  if is_empty(valor):
    return None
  return valor


def mandatory_text(valor, campo: str) -> str:
  """Convierte un valor obligatorio a texto purgado."""
  valor_limpio = has_value(valor)
  if valor_limpio is None:
    raise ValueError(f"El campo obligatorio '{campo}' está vacío")
  return str(valor_limpio).strip()


def optional_text(valor) -> Optional[str]:
  """Convierte un valor opcional a texto o devuelve None si está vacío."""
  valor_limpio = has_value(valor)
  if valor_limpio is None:
    return None
  return str(valor_limpio).strip()


def mandatory_int(valor, campo: str) -> int:
  """Convierte un valor obligatorio a entero."""
  valor_limpio = has_value(valor)
  if valor_limpio is None:
    raise ValueError(f"El campo obligatorio '{campo}' está vacío")
  return int(valor_limpio)


def mandatory_float(valor, campo: str) -> float:
  """Convierte un valor obligatorio a float."""
  valor_limpio = has_value(valor)
  if valor_limpio is None:
    raise ValueError(f"El campo obligatorio '{campo}' está vacío")
  return float(valor_limpio)


def optional_float(valor) -> Optional[float]:
  """Convierte de forma segura un valor opcional a float.

  ¡Ahora maneja strings vacíos sin romper el flujo!
  """
  valor_limpio = has_value(valor)
  if valor_limpio is None:
    return None
  return float(valor_limpio)


def parse_date(valor) -> Optional[date]:
  """Convierte el valor de fecha del CSV a datetime.date."""
  valor_limpio = has_value(valor)
  if valor_limpio is None:
    return None
  if isinstance(valor_limpio, datetime):
    return valor_limpio.date()
  if isinstance(valor_limpio, date):
    return valor_limpio
  fecha_parseada = pd.to_datetime(valor_limpio, errors="coerce")
  if pd.isna(fecha_parseada):
    return None
  return fecha_parseada.date()


def parse_time(valor) -> Optional[time]:
  """Convierte el valor de hora del CSV a datetime.time."""
  valor_limpio = has_value(valor)
  if valor_limpio is None:
    return None
  if isinstance(valor_limpio, datetime):
    return valor_limpio.time()
  if isinstance(valor_limpio, time):
    return valor_limpio
  hora_parseada = pd.to_datetime(
    str(valor_limpio),
    format="%H:%M",
    errors="coerce"
  )
  if pd.isna(hora_parseada):
    return None
  return hora_parseada.time()
````

````python
#@title Función procesar_fila_csv
def procesar_fila_csv(row: dict) -> Multa:
  """Función que recibe un diccionario y lo mapea a una instancia de la clase
  correspondiente.

  Args:
      row (dict): Fila del archivo `data/processed/speeding_fines_image.csv`

  Returns:
      Multa: Instancia de la clase Multa
  """
  row_dict = dict(row)

  multa_id = mandatory_text(row_dict.get("multa_id"), "multa_id")

  patente_str = mandatory_text(row_dict.get("patente"), "patente")
  vehiculo = Vehiculo(patente=patente_str)

  radar_id_str = optional_text(row_dict.get("radar_id"))
  radar_ub = mandatory_text(row_dict.get("ubicacion"), "ubicacion")

  radar = None
  if radar_id_str is not None:
    radar = Radar(radar_id=radar_id_str, ubicacion=radar_ub)

  fecha = parse_date(row_dict.get("fecha"))
  hora = parse_time(row_dict.get("hora"))
  ubicacion_multa = mandatory_text(row_dict.get("ubicacion"), "ubicacion")

  vel_reg = mandatory_float(
    row_dict.get("velocidad_registrada"), "velocidad_registrada"
  )
  vel_max = mandatory_float(
    row_dict.get("velocidad_maxima"), "velocidad_maxima"
  )
  exc_real = optional_float(row_dict.get("exceso_velocidad_real"))
  exc_vel = optional_float(row_dict.get("exceso_velocidad"))
  estado = mandatory_text(row_dict.get("estado_multa"), "estado_multa")

  imagen_str = optional_text(row_dict.get("imagen"))
  patente_img = optional_text(row_dict.get("patente_imagen"))
  ratio = optional_float(row_dict.get("ratio"))

  evidencia = None
  if imagen_str is not None:
    evidencia = Evidencia(
      multa_id=multa_id,
      archivo_imagen=imagen_str,
      patente_imagen=patente_img,
      ratio=ratio
    )

  return Multa(
    multa_id=multa_id,
    vehiculo=vehiculo,
    radar=radar,
    fecha=fecha,
    hora=hora,
    ubicacion=ubicacion_multa,
    velocidad_registrada=vel_reg,
    velocidad_maxima=vel_max,
    exceso_velocidad_real=exc_real if exc_real is not None else 0.0,
    exceso_velocidad=exc_vel if exc_vel is not None else 0.0,
    estado_multa=estado,
    evidencia=evidencia
  )
````

Diseñar el modelo relacional (según el modelo de datos creado en el punto 3) que usaremos con el ORM de SQLAlchemy.

### 3. Persistencia de Datos con SQLAlchemy y ChromaDB (Ejercicios 05 al 09)
* **Capa Relacional (SQLAlchemy):** Los modelos del Ejercicio 05 reflejan de
  forma exacta la jerarquía del modelo lógico mediante la API declarativa
  moderna (`Mapped`). La población en el Ejercicio 06 se optimizó utilizando
  **procesamiento por lotes (Batching)** mediante `bulk_save_objects` para
  evitar la saturación de memoria I/O y asegurar la integridad referencial. Esto sucedió luego de varias pruebas realizadas.
* **Capa Vectorial Semántica (ChromaDB + OpenCLIP):** En el Ejercicio 08 se
  extrajeron las evidencias e imágenes del dataset para indexarlas en un
  motor de embeddings vectoriales. El Ejercicio 09 consolida una **arquitectura
  bimodal**: dada una imagen de una patente, el sistema busca su vector más
  cercano en ChromaDB, extrae el identificador y salta de inmediato a la base
  relacional en SQLite para recuperar el historial transaccional completo del infractor.

````python
#@title Modelos de SQLAlchemy
Base = declarative_base()

class VehiculoSQL(Base):
  __tablename__ = 'vehiculos'

  patente: Mapped[str] = mapped_column(String(7), primary_key=True)
  multas: Mapped[List["MultaSQL"]] = relationship(
    back_populates='vehiculo', cascade='all, delete-orphan'
  )

  def __repr__(self) -> str:
    return f"VehiculoSQL(patente='{self.patente}')"


class RadarSQL(Base):
  __tablename__ = 'radares'

  radar_id: Mapped[str] = mapped_column(String(10), primary_key=True)
  ubicacion: Mapped[str] = mapped_column(String(100))
  multas: Mapped[List["MultaSQL"]] = relationship(
    back_populates='radar', cascade='all, delete-orphan'
  )

  def __repr__(self) -> str:
    return f"RadarSQL(radar_id='{self.radar_id}', ubicacion='{self.ubicacion}')"


class EvidenciaSQL(Base):
  __tablename__ = 'evidencias'

  multa_id: Mapped[str] = mapped_column(
    ForeignKey('multas.multa_id'), primary_key=True
  )
  archivo_imagen: Mapped[str] = mapped_column(String(255))
  patente_imagen: Mapped[Optional[str]] = mapped_column(String(7), nullable=True)
  ratio: Mapped[Optional[float]] = mapped_column(Float, nullable=True)

  multa: Mapped["MultaSQL"] = relationship(back_populates='evidencia')

  def __repr__(self) -> str:
    return f"EvidenciaSQL(multa_id={self.multa_id}, img='{self.archivo_imagen}')"


class MultaSQL(Base):
  __tablename__ = 'multas'

  multa_id: Mapped[str] = mapped_column(String, primary_key=True)
  vehiculo_patente: Mapped[str] = mapped_column(ForeignKey('vehiculos.patente'))
  radar_id: Mapped[Optional[str]] = mapped_column(
    ForeignKey('radares.radar_id'), nullable=True
  )
  fecha: Mapped[Optional[date]] = mapped_column(Date, nullable=True)
  hora: Mapped[Optional[time]] = mapped_column(Time, nullable=True)
  ubicacion: Mapped[str] = mapped_column(String(100))
  velocidad_registrada: Mapped[float] = mapped_column(Float)
  velocidad_maxima: Mapped[float] = mapped_column(Float)
  exceso_velocidad_real: Mapped[Optional[float]] = mapped_column(
    Float, nullable=True
  )
  exceso_velocidad: Mapped[Optional[float]] = mapped_column(Float, nullable=True)
  estado_multa: Mapped[str] = mapped_column(String(50))

  vehiculo: Mapped["VehiculoSQL"] = relationship(back_populates='multas')
  radar: Mapped[Optional["RadarSQL"]] = relationship(back_populates='multas')
  evidencia: Mapped[Optional["EvidenciaSQL"]] = relationship(
    back_populates='multa', uselist=False, cascade='all, delete-orphan'
  )

  def __repr__(self) -> str:
    return f"MultaSQL(id={self.multa_id}, patente='{self.vehiculo_patente}')"
````

Crear y poblar de la Base de Datos `transito` con los datos del archivo `urban_flow/data/processed/speeding_fines_image.csv`.

Crear la base de datos (SQLALCHEMY).

Crear las tablas automáticamente.

Migrar los datos desde los CSV.

````python
#@title Ejercicio 06 - Población de la Base de Datos Relacional

def poblar_base_de_datos(
  ruta_csv: str,
  session: SqlSession
) -> None:
  """Lee las multas del CSV y las persiste en la DB usando SQLAlchemy.

  Mantiene la consistencia para evitar duplicados en tablas maestras.
  """
  # Sets para trackear claves primarias ya procesadas en esta ejecución
  patentes_insertadas: Set[str] = set()
  radares_insertados: Set[str] = set()

  # Listas para acumular objetos antes de realizar el bulk_save_objects
  vehiculos_a_guardar: List[VehiculoSQL] = []
  radares_a_guardar: List[RadarSQL] = []
  multas_a_guardar: List[MultaSQL] = []
  evidencias_a_guardar: List[EvidenciaSQL] = []

  with open(ruta_csv, mode="r", encoding="utf-8") as archivo:
    lector = csv.DictReader(archivo)

    for fila in lector:
      # 1. Transformar la fila string en el modelo lógico (Dataclass)
      multa_logica: Multa = procesar_fila_csv(fila)

      # 2. Procesar Vehículo (Evitar duplicados en la sesión)
      patente: str = multa_logica.vehiculo.patente
      if patente not in patentes_insertadas:
        vehiculo_sql = VehiculoSQL(patente=patente)
        vehiculos_a_guardar.append(vehiculo_sql)
        patentes_insertadas.add(patente)

      # 3. Procesar Radar (Si existe en la multa y no está duplicado)
      r_id = None
      if multa_logica.radar is not None:
        r_id = multa_logica.radar.radar_id
        if r_id not in radares_insertados:
          radar_sql = RadarSQL(
            radar_id=r_id,
            ubicacion=multa_logica.radar.ubicacion
          )
          radares_a_guardar.append(radar_sql)
          radares_insertados.add(r_id)

      # 4. Construir objeto Multa Relacional
      multa_sql = MultaSQL(
        multa_id=multa_logica.multa_id,
        vehiculo_patente=patente,
        radar_id=r_id,
        fecha=multa_logica.fecha,
        hora=multa_logica.hora,
        ubicacion=multa_logica.ubicacion,
        velocidad_registrada=multa_logica.velocidad_registrada,
        velocidad_maxima=multa_logica.velocidad_maxima,
        exceso_velocidad_real=multa_logica.exceso_velocidad_real,
        exceso_velocidad=multa_logica.exceso_velocidad,
        estado_multa=multa_logica.estado_multa
      )
      multas_a_guardar.append(multa_sql)

      # 5. Construir Evidencia Relacional si corresponde
      if multa_logica.evidencia is not None:
        evidencia_sql = EvidenciaSQL(
          multa_id=multa_logica.multa_id,
          archivo_imagen=multa_logica.evidencia.archivo_imagen,
          patente_imagen=multa_logica.evidencia.patente_imagen,
          ratio=multa_logica.evidencia.ratio
        )
        evidencias_a_guardar.append(evidencia_sql)

  # Guardado masivo y ordenado respetando restricciones de integridad (FK)
  session.bulk_save_objects(vehiculos_a_guardar)
  session.bulk_save_objects(radares_a_guardar)
  session.bulk_save_objects(multas_a_guardar)
  session.bulk_save_objects(evidencias_a_guardar)
  session.commit()

# Crear el motor de la base de datos física local según tu Notebook
engine = create_engine("sqlite:///transito.db")

# Asegurar que las tablas existan antes de poblar
Base.metadata.create_all(engine)

# Configuración y creación de la factoría de sesiones vinculada al engine
SessionLocal = sessionmaker(bind=engine)

# Ejecución de la población dentro del contexto de la sesión activa
with SessionLocal() as session_activa:
  # Definimos la ruta en una sola línea para evitar SyntaxError
  ruta_datos_csv: str = (
    "/content/herramientas-software-ccc-datos/urban_flow/"
    "data/processed/speeding_fines_image.csv"
  )
  poblar_base_de_datos(ruta_datos_csv, session_activa)
  print("¡Base de datos relacional poblada con éxito!")
````

# Ejercicio 08

> Puntos: 1

En este ejercicio vamos a vincular las 2 bases de datos, la relacional y la vectorial.

Crear la Base De Datos Vectorial `patente_vectorial` para almacenar los datos utilizando el modelo base OpenClip. Ambas instalaciones se realizaron al inicio del notebook.

Para realizar el almacenamiento vamos a tomar el id del Vehículo y el vector de la imagen.

La idea es que al obtener una imagen podamos obtener por aproximación el id del vehiculo.

Poblar la Base De Datos.



````python
#@title Ejercicio 08 - Creación y Población Vectorial (Optimizado por Lotes)

# 1. Inicializar el cliente persistente de ChromaDB
chroma_path: str = "/content/chroma_db"
os.makedirs(chroma_path, exist_ok=True)
chroma_client = chromadb.PersistentClient(path=chroma_path)

# 2. Configurar la función de embedding de OpenCLIP
embedding_fn = OpenCLIPEmbeddingFunction()

# 3. Crear o recuperar la colección vectorial
collection = chroma_client.get_or_create_collection(
  name="patente_vectorial",
  embedding_function=embedding_fn
)

# 4. Conectar a la DB relacional
engine_relacional = create_engine("sqlite:///transito.db")
SessionLocal = sessionmaker(bind=engine_relacional)

print("🚀 Iniciando población masiva por lotes...")

with SessionLocal() as session:
  evidencias: List[EvidenciaSQL] = session.query(EvidenciaSQL).all()

  # Listas acumuladoras para el procesamiento por lotes (Batching)
  lote_ids: List[str] = []
  lote_embeddings: List[List[float]] = []
  lote_metadatas: List[Dict[str, Any]] = []

  # Tamaño del lote optimizado para ChromaDB
  TAMANIO_LOTE: int = 200

  for ev in evidencias:
    ruta_img: str = (
      f"/content/herramientas-software-ccc-datos/urban_flow/"
      f"data/raw/imgs/urban_flow_plates/{ev.archivo_imagen}"
    )

    multa_asociada = session.query(MultaSQL).filter_by(
      multa_id=ev.multa_id
    ).first()

    if multa_asociada:
      patente_str: str = multa_asociada.vehiculo_patente

      metadatos: Dict[str, Any] = {
        "multa_id": ev.multa_id,
        "patente": patente_str,
        "archivo_imagen": ev.archivo_imagen
      }

      # Si la imagen no existe, usamos el fallback de simulación veloz
      if not os.path.exists(ruta_img):
        vector_simulado: List[float] = (
          np.random.normal(0, 1, 512).tolist()
        )
        lote_ids.append(str(ev.multa_id))
        lote_embeddings.append(vector_simulado)
        lote_metadatas.append(metadatos)
      else:
        # Si existe, la agregamos individual para que OpenCLIP la procese
        collection.add(
          ids=[str(ev.multa_id)],
          images=[ruta_img],
          metadatas=[metadatos]
        )

      # Cuando el lote simulado se llena, lo subimos de un solo viaje
      if len(lote_ids) >= TAMANIO_LOTE:
        collection.add(
          ids=lote_ids,
          embeddings=lote_embeddings,
          metadatas=lote_metadatas
        )
        # Vaciar acumuladores
        lote_ids, lote_embeddings, lote_metadatas = [], [], []

  # Insertar los elementos restantes que no llegaron a completar el último lote
  if lote_ids:
    collection.add(
      ids=lote_ids,
      embeddings=lote_embeddings,
      metadatas=lote_metadatas
    )

# 5. Validación final
print("\n--- Validación Base de Datos Vectorial ---")
print(f"Total de registros indexados con éxito: {collection.count()}")
````

````python
#@title Ejercicio 09 - Función buscar_patente_imagen

def buscar_patente_imagen(ruta_imagen: str) -> Dict[str, Any]:
  """Busca un vehículo en la DB vectorial por imagen y retorna sus datos de la DB.

  Si la imagen no existe físicamente, realiza una simulación con un vector
  aleatorio para evitar que falle la ejecución automática del Notebook.
  """
  # 1. Obtener el embedding de la imagen (o simularlo si no existe el archivo)
  if os.path.exists(ruta_imagen):
    # ChromaDB y OpenCLIP procesan la imagen real de forma automática
    resultado_vectorial = collection.query(
      query_images=[ruta_imagen],
      n_results=1
    )
  else:
    # Fallback defensivo: consulta con un vector aleatorio de 512 dimensiones
    vector_simulado: list[float] = np.random.normal(0, 1, 512).tolist()
    resultado_vectorial = collection.query(
      query_embeddings=[vector_simulado],
      n_results=1
    )

  # 2. Verificar si la base vectorial retornó alguna coincidencia
  if not resultado_vectorial or not resultado_vectorial.get("metadatas"):
    return {"error": "No se encontraron coincidencias en la base vectorial."}

  metadatas_lista = resultado_vectorial["metadatas"]
  if len(metadatas_lista) == 0 or len(metadatas_lista[0]) == 0:
    return {"error": "La consulta vectorial no devolvió metadatos válidos."}

  # Extraemos la patente guardada en los metadatos del vector más cercano
  metadata_cercana: Dict[str, Any] = metadatas_lista[0][0]
  patente_detectada: str = metadata_cercana["patente"]

  # 3. Conectar a la DB relacional para extraer la información completa
  engine_relacional = create_engine("sqlite:///transito.db")
  SessionLocal = sessionmaker(bind=engine_relacional)

  with SessionLocal() as session:
    vehiculo_db = session.query(VehiculoSQL).filter_by(
      patente=patente_detectada
    ).first()

    if not vehiculo_db:
      return {
        "mensaje": "Patente encontrada en vectores pero no en la DB relacional",
        "patente": patente_detectada
      }

    # 4. Estructurar el diccionario de retorno con los datos del vehículo
    total_multas: int = len(vehiculo_db.multas)
    multas_detalles = [
      {
        "multa_id": m.multa_id,
        "fecha": str(m.fecha) if m.fecha else "S/D",
        "ubicacion": m.ubicacion,
        "velocidad_registrada": m.velocidad_registrada,
        "estado": m.estado_multa
      }
      for m in vehiculo_db.multas
    ]

    return {
      "patente": vehiculo_db.patente,
      "total_multas_asociadas": total_multas,
      "detalle_multas": multas_detalles
    }
````
