- Desarrollar un función para mostrar 8 imágenes de forma aleatorea en forma de tabla de 2 columnas y 4 filas. Debe mostrar 4 imágenes de cada grupo. Es una función reutilizable. En este punto se deben mostrar las imágenes originales. Cada imágen debe tener un ancho de 10 pulgas y un alto de 8 pulgadas, se debe mostrar el nombre de la imagen.



````python
#@title Listado de imágenes y creación de grupos plates y completes.
# 1. Listar imágenes y tamaños (NO SE IMPRIME en pantalla según consigna)
images_path: str = "/content/herramientas-software-ccc-datos/urban_flow/data/raw/imgs/urban_flow_plates"

# Almacenamos (nombre, tamaño_kb) en una lista de tuplas con tipado estricto
all_images_files: List[Tuple[str, float]] = [
  (f, os.path.getsize(os.path.join(images_path, f)) / 1024)
  for f in os.listdir(images_path)
]

# 2. Separar en 2 grupos utilizando el diccionario solicitado
group_images: Dict[str, List[Dict[str, Any]]] = {
  "plates": [],
  "completes": []
}

total_pixels: int = 0
valid_images_count: int = 0

for filename, size_kb in all_images_files:
  full_path: str = os.path.join(images_path, filename)

  # Usamos cv2 ya que está aprobado por la cátedra
  img: Any = cv2.imread(full_path)

  if img is not None:
    height, width, _ = img.shape
    area: int = width * height
    total_pixels += area
    valid_images_count += 1

    img_data: Dict[str, Any] = {
      "filename": filename,
      "width": width,
      "height": height,
      "area": area,
      "path": full_path,
      "patent": ""
    }

    # Lógica de clasificación por nombre o por umbral de tamaño
    if "patente" in filename.lower() or area < 120000:
      group_images["plates"].append(img_data)
    else:
      group_images["completes"].append(img_data)

# Calcular la resolución promedio (Ayuda de la consigna)
avg_resolution: float = total_pixels / valid_images_count if valid_images_count > 0 else 0.0

# Salida exacta solicitada por la consigna
print(f"El grupo 'plates' contiene {len(group_images['plates'])} imágenes.")
print(f"El grupo 'completes' contiene {len(group_images['completes'])} imágenes.")
print(f"La resolución promedio del dataset es de: {avg_resolution:,.2f} píxeles.")

# Guardar el archivo JSON interim
output_dir: str = "/content/herramientas-software-ccc-datos/urban_flow/data/interim"
os.makedirs(output_dir, exist_ok=True)
with open(os.path.join(output_dir, "group_images.json"), "w") as f:
  json.dump(group_images, f, indent=2)

!git status
!git add urban_flow/data/interim/group_images.json
!git commit -m 'feat: Add group_images.json'
git_sync('sprint_2', activar_git_push)
````

````python
#@title Definición función random_images
def show_random_images(groups_dict: Dict[str, List[Dict[str, Any]]]) -> None:
  """Selecciona aleatoriamente 4 imágenes de cada grupo y las muestra.

  Muestra las imágenes en una cuadrícula de 4 filas y 2 columnas.
  Configura el tamaño de la figura para que cada celda mida aprox 10x8 pulgadas.
  """
  selected_plates: List[Dict[str, Any]] = random.sample(groups_dict["plates"], 4)
  selected_completes: List[Dict[str, Any]] = random.sample(groups_dict["completes"], 4)

  # 2 columnas, 4 filas. Para que cada una sea de 10x8
  fig, axes = plt.subplots(4, 2, figsize=(10, 8))

  for row in range(4):
    # Columna 1: Plates
    plate_data = selected_plates[row]
    img_p = cv2.imread(plate_data["path"])
    img_p_rgb = cv2.cvtColor(img_p, cv2.COLOR_BGR2RGB)
    axes[row, 0].imshow(img_p_rgb)
    axes[row, 0].set_title(f"Plate: {plate_data['filename']}", fontsize=12)
    axes[row, 0].axis("off")

    # Columna 2: Completes
    complete_data = selected_completes[row]
    img_c = cv2.imread(complete_data["path"])
    img_c_rgb = cv2.cvtColor(img_c, cv2.COLOR_BGR2RGB)
    axes[row, 1].imshow(img_c_rgb)
    axes[row, 1].set_title(f"Complete: {complete_data['filename']}", fontsize=12)
    axes[row, 1].axis("off")

  plt.tight_layout()
  plt.show()

# Invocación de la función
show_random_images(group_images)
````

# Ejercicio 03

> puntos: 2

En este punto realizaremos un procesamiento de las imágenes, en cada punto se debe llamar a la función que muestra las imágenes trabajadas en el punto anterior.

- Conversión a escala de grises de las imágenes originales y almacenarlas en `urban_flow/data/interim/imgs/03_01_gray_scale/plates` y `urban_flow/data/interim/imgs/03_01_gray_scale/completes`, según la clasificación del ejercicio 2.   

- Realizar un suavizado de las imágenes en escala de grises y almacenarlas en `urban_flow/data/interim/imgs/03_02_blur` según la clasificación del ejercicio 2.

- Realizar una detección de bordes sobre las imágenes suavizadas para intentar detectar las patentes y almacenarlas en `urban_flow/data/interim/imgs/03_03_canny` según la clasificación del ejercicio 2.



````python
#@title Procesamiento de imágenes: Escala de Grises, Suavizado y Canny
def procesar_imagenes_sprint2(
  group_images: Dict[str, List[Dict[str, Any]]]
) -> Tuple[Dict[str, List[str]], Dict[str, List[str]], Dict[str, List[str]]]:
  """Realiza el procesamiento de imágenes: Escala de grises, Suavizado y Canny.

  Crea las estructuras de directorios requeridas, procesa cada imagen según
  su clasificación del ejercicio anterior y guarda los resultados.

  Args:
    group_images: Diccionario con la clasificación de 'plates' y 'completes'.

  Returns:
    Tres diccionarios con las rutas de las imágenes procesadas en cada etapa.
  """
  # 1. Definición de rutas solicitadas por la cátedra (cortadas a 80 caracteres)
  base_dir: str = "/content/herramientas-software-ccc-datos/urban_flow/data"

  path_gray: Dict[str, str] = {
    "plates": f"{base_dir}/interim/imgs/03_01_gray_scale/plates",
    "completes": f"{base_dir}/interim/imgs/03_01_gray_scale/completes"
  }
  path_blur: Dict[str, str] = {
    "plates": f"{base_dir}/interim/imgs/03_02_blur/plates",
    "completes": f"{base_dir}/interim/imgs/03_02_blur/completes"
  }
  path_canny: Dict[str, str] = {
    "plates": f"{base_dir}/interim/imgs/03_03_canny/plates",
    "completes": f"{base_dir}/interim/imgs/03_03_canny/completes"
  }

  # Creación automática de directorios usando os.makedirs
  for folder_path in (list(path_gray.values()) +
                      list(path_blur.values()) +
                      list(path_canny.values())):
    os.makedirs(folder_path, exist_ok=True)

  # Diccionarios de salida para almacenar las rutas de los archivos generados
  rutas_gray: Dict[str, List[str]] = {"plates": [], "completes": []}
  rutas_blur: Dict[str, List[str]] = {"plates": [], "completes": []}
  rutas_canny: Dict[str, List[str]] = {"plates": [], "completes": []}

  # Iteración sobre los dos grupos definidos en el Sprint anterior
  for grupo in ["plates", "completes"]:
    for img_info in group_images[grupo]:
      filename: str = img_info["filename"]
      orig_path: str = img_info["path"]

      # Lectura de la imagen original usando OpenCV
      img: Any = cv2.imread(orig_path)
      if img is None:
        continue

      # --- Paso 1: Conversión a Escala de Grises ---
      img_gray: Any = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
      gray_dest: str = os.path.join(path_gray[grupo], filename)
      cv2.imwrite(gray_dest, img_gray)
      rutas_gray[grupo].append(gray_dest)

      # --- Paso 2: Suavizado (Gaussian Blur) ---
      # Usamos un kernel de 5x5 recomendado para remoción de ruido previo a Canny
      img_blur: Any = cv2.GaussianBlur(img_gray, (5, 5), 0)
      blur_dest: str = os.path.join(path_blur[grupo], filename)
      cv2.imwrite(blur_dest, img_blur)
      rutas_blur[grupo].append(blur_dest)

      # --- Paso 3: Detección de Bordes (Canny) ---
      # Umbrales estándar 100 y 200 para detección de bordes de la patente
      img_canny: Any = cv2.Canny(img_blur, 100, 200)
      canny_dest: str = os.path.join(path_canny[grupo], filename)
      cv2.imwrite(canny_dest, img_canny)
      rutas_canny[grupo].append(canny_dest)

  return rutas_gray, rutas_blur, rutas_canny


def mostrar_imagenes_procesadas(
  rutas_dict: Dict[str, List[str]],
  titulo_etapa: str
) -> None:
  """Muestra de forma aleatoria 4 imágenes de cada grupo para una etapa dada.

  Args:
    rutas_dict: Diccionario que contiene las rutas de los archivos procesados.
    titulo_etapa: Texto que identifica la fase de procesamiento a mostrar.
  """
  # Selección aleatoria de 4 imágenes por grupo (evita errores si hay menos)
  num_muestras: int = min(4, len(rutas_dict["plates"]), len(rutas_dict["completes"]))
  sel_plates: List[str] = random.sample(rutas_dict["plates"], num_muestras)
  sel_completes: List[str] = random.sample(rutas_dict["completes"], num_muestras)

  # Configuración de tamaño de figura acorde a especificaciones de 10x8 por celda
  fig, axes = plt.subplots(num_muestras, 2, figsize=(20, num_muestras * 8))
  fig.suptitle(f"Visualización de Etapa: {titulo_etapa}", fontsize=16)

  for i in range(num_muestras):
    # Procesamiento y muestra del grupo Plates (Columna 1)
    img_p: Any = cv2.imread(sel_plates[i])
    if len(img_p.shape) == 3:
      img_p = cv2.cvtColor(img_p, cv2.COLOR_BGR2RGB)
    axes[i, 0].imshow(img_p, cmap="gray" if len(img_p.shape) == 2 else None)
    axes[i, 0].set_title(f"Plate: {os.path.basename(sel_plates[i])}")
    axes[i, 0].axis("off")

    # Procesamiento y muestra del grupo Completes (Columna 2)
    img_c: Any = cv2.imread(sel_completes[i])
    if len(img_c.shape) == 3:
      img_c = cv2.cvtColor(img_c, cv2.COLOR_BGR2RGB)
    axes[i, 1].imshow(img_c, cmap="gray" if len(img_c.shape) == 2 else None)
    axes[i, 1].set_title(f"Complete: {os.path.basename(sel_completes[i])}")
    axes[i, 1].axis("off")

  plt.tight_layout()
  plt.show()

# --- Ejecución de las funciones y visualizaciones interactivas ---
r_gray, r_blur, r_canny = procesar_imagenes_sprint2(group_images)

# La consigna pide llamar a la función que muestra las imágenes trabajadas
mostrar_imagenes_procesadas(r_gray, "03_01_escala_de_grises")
mostrar_imagenes_procesadas(r_blur, "03_02_suavizado_blur")
mostrar_imagenes_procesadas(r_canny, "03_03_bordes_canny")

!git status
!git add urban_flow/data/interim/imgs/
!git commit -m 'feat: Add parsed images'
git_sync('sprint_2', activar_git_push)
````

````python
#@title Definición de la función extraer_patente

import easyocr
def extraer_patente(image_path) -> str:
  """ Extrae la patente de una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Patente extraída.
  """
  reader = easyocr.Reader(['es'])
  detection:list = reader.readtext(image_path)
  if len(detection) == 0:
    return ''
  patent = ''
  for d in detection:
    if isinstance(d,tuple) and len(d) == 3:
      if isinstance(d[1],str):
        patent+=d[1]
  return patent
````

````python
#@title Definición de la función match_strings
def match_strings(detected: str, original: str) -> Tuple[bool, float]:
  """Calcula la coincidencia de caracteres de izquierda a derecha de 2 strings.

  Devuelve si supera el umbral del 80% y el ratio calculado.
  """
  if not detected or not original:
    return False, 0.0

  # Limpieza básica para estandarizar la comparación
  det: str = "".join(c for c in detected if c.isalnum()).upper()
  orig: str = "".join(c for c in original if c.isalnum()).upper()

  longitud_minima: int = min(len(det), len(orig))
  if longitud_minima == 0:
    return False, 0.0

  coincidencias: int = 0
  for i in range(longitud_minima):
    if det[i] == orig[i]:
      coincidencias += 1

  ratio: float = coincidencias / max(len(det), len(orig))
  return ratio >= 0.8, ratio
````

````python
#@title Procesamiento visual y comparación

# 1. Extracción de patentes desde las imágenes en escala de grises
# Usamos escala de grises en vez de Canny para optimizar la lectura del OCR
base_gray_path: str = "/content/herramientas-software-ccc-datos/urban_flow/data/interim/imgs/03_01_gray_scale"

for grupo in ["plates", "completes"]:
  for data in group_images[grupo]:
    input_path: str = os.path.join(base_gray_path, grupo, data["filename"])

    # Si por alguna razón no existiera el interim, recurre al original
    if not os.path.exists(input_path):
      input_path = data["path"]

    txt_detectado: str = extraer_patente(input_path)
    data["patent"] = txt_detectado

# 2. Guardar el estado del diccionario actualizado en el interim JSON
interim_dir: str = "/content/herramientas-software-ccc-datos/urban_flow/data/interim"
with open(os.path.join(interim_dir, "group_images.json"), "w") as f:
  json.dump(group_images, f, indent=2)

# 3. Carga de Dataset Speeding Fines y cruce de datos
df_fines: pd.DataFrame = pd.read_csv(os.path.join(interim_dir, "speeding_fines.csv"))
plates_list_unique: List[str] = df_fines["patente"].dropna().unique().tolist()

map_imagen: Dict[str, str] = {}
map_patente_imagen: Dict[str, str] = {}
map_ratio: Dict[str, float] = {}

# Algoritmo de emparejamiento basado en la lógica solicitada
for grupo, lista_imagenes in group_images.items():
  for img_data in lista_imagenes:
    search_plate: str = img_data["patent"]
    if not search_plate:
      continue

    for original_plate in plates_list_unique:
      es_match, ratio_calculado = match_strings(search_plate, original_plate)
      if es_match:
        # Si la patente ya existía, se conserva la que tenga mayor porcentaje
        if original_plate not in map_ratio or ratio_calculado > map_ratio[original_plate]:
          map_imagen[original_plate] = img_data["path"]
          map_patente_imagen[original_plate] = search_plate
          map_ratio[original_plate] = ratio_calculado

# 4. Mapeo y generación de columnas nuevas
df_fines["imagen"] = df_fines["patente"].map(map_imagen)
df_fines["patente_imagen"] = df_fines["patente"].map(map_patente_imagen)
df_fines["ratio"] = df_fines["patente"].map(map_ratio)

# 5. Almacenamiento del archivo final procesado
processed_dir: str = "/content/herramientas-software-ccc-datos/urban_flow/data/processed"
os.makedirs(processed_dir, exist_ok=True)
df_fines.to_csv(os.path.join(processed_dir, "speeding_fines_image.csv"), index=False)

# 6. Validaciones finales en pantalla
df_matches: pd.DataFrame = df_fines[df_fines["ratio"].notna()]
print(f"Total de filas que hicieron match exitoso: {len(df_matches)}")

!git status
!git add urban_flow/data/processed/speeding_fines_image.csv
!git commit -m 'feat: Add speeding_fines_image.csv'
git_sync('sprint_2', activar_git_push)
````
