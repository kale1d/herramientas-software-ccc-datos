Normalización de fechas

````python
# Normalizar las de fechas con el formato 'YYYY-MM-DD', en caso de fechas inválidas completar con la fecha 1932-01-01.
#  Mostrar las 10 primeras fechas.

df_fines['fecha'] = pd.to_datetime(df_fines['fecha'], errors='coerce')
df_fines['fecha'] = df_fines['fecha'].fillna(pd.Timestamp('1932-01-01'))

print("--- 10 primeras fechas ---")
df_fines['fecha'].head(10)
````

Normalización de horas

````python
# Normalizar los horas al formato de 24hs, las horas inválidas se procesaran como 00:00. Mostrar las 10 primeras horas.
# Saber la cantidad de celdas que tienen invalid hour.

cant_invalidos = (df_fines['hora']=='invalid_hour').sum()
print(f"Cantidad de invalidos {cant_invalidos}")

#Sacar las horas invalidas
df_fines['hora'] = df_fines['hora'].replace('invalid_hour', '00:00')

#Formatear a 24 horas
df_fines['hora'] = pd.to_datetime(df_fines['hora'], errors='coerce').dt.strftime('%H:%M')
print("--- 10 primeras horas ---")
print(df_fines['hora'].head(10))
````

Normalización de ubicaciones

````python
# Normalizar las ubicaciones. Quitar caracteres especiales y pasar todo a mayúsculas. Mostrar las 10 primeras ubicaciones.

# Antes de hacer la limpieza se valida/busca si hay caracteres especiales en el dataset
texto = "".join(df_fines['ubicacion'].dropna().astype(str))
char_especiales= set(re.findall(r'[^a-zA-Z0-9\s]', texto))
print(f"Caracteres especiales encontrados: {char_especiales}")

#De la búsqueda sale que unicamente hay '.' como caracteres especiales por lo cual la quita es mas simple
df_fines['ubicacion'] = df_fines['ubicacion'].str.replace(r'\.', '', regex=True).str.strip().str.upper()

print("--- 10 primeras ubicaciones ---")
print(df_fines['ubicacion'].head(10))
````

Limpieza y normalización de patentes

````python
# Limpiar y normalizar las patentes, en caso de error completar con pd.NA (no disponible/not available).
#  Quitar caracteres especiales y pasar todo a mayúsculas. Mostrar las 10 últimas patentes.
#Visualizar patentes especiales

df_fines['patente'].unique()
df_fines['patente'] = df_fines['patente'].str.replace(r'[^a-zA-Z0-9]', '', regex=True).str.strip().str.upper()
df_fines['patente'].isnull().sum()
df_fines.loc[df_fines['patente'] == '', 'patente'] = pd.NA
df_fines.loc[df_fines['patente'] == 'NaN', 'patente'] = pd.NA

#df_fines.tail(10)
print("--- 10 últimas patentes ---")
print(df_fines['patente'].tail(10))
````

Eliminación de valores vacíos

````python
# Eliminar las filas que posean valores relevantes para las multas y que se encuentren vacios.
# Mostrar la cantidad de registros eliminados y cuales son las columnas que se observaron utilizando la leyenda
#Buscar la cantidad de valores nulos que hay en velocidad_registrada

vel_max_nulos=df_fines['velocidad_maxima'].isna().sum()
print(f"Cantidad de valores nulos en velocidad_maxima: {vel_max_nulos}")

vel_nulos=df_fines['velocidad_registrada'].isna().sum()
print(f"Cantidad de valores nulos en velocidad_registrada: {vel_nulos}")

patentes_sin_limpiar_vel =df_fines['patente'].isnull().sum()
print(f"Cantidad de patentes nulas antes de limpiar: {patentes_sin_limpiar_vel}")

df_limpio = df_fines.dropna(subset=['velocidad_registrada', 'patente'])
filas_sin_limpiar=len(df_fines)
filas_limpias=len(df_limpio)
filas_eliminadas= int = filas_sin_limpiar - filas_limpias
print(f"Cantidad de filas sin limpiar: {filas_sin_limpiar}")
print(f"Cantidad de filas limpias: {filas_limpias}")

print(f"Se eliminaron {filas_eliminadas} filas.")
print("Las columnas observadas son: Velocidad Registrada y Patentes.")
print("Si alguno de esos dos valores no existe no se puede validar que la multa sea correcta")

# Se eliminaron 1982 filas.
# Las columnas observadas son:
# - velocidad_registrada
# - patentes
# Si alguno de estos dos valores no existe no se puede validar que la multa sea correcta.
````

Detección de Outliers (Valores Atípicos)

````python
# Detectar y eliminar outliers en velocidad_registrada usando el método IQR

Q1 = df_limpio['velocidad_registrada'].quantile(0.25)
Q3 = df_limpio['velocidad_registrada'].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Filtrar el dataframe
df_sin_outliers = df_limpio[(df_limpio['velocidad_registrada'] >= limite_inferior) &
                             (df_limpio['velocidad_registrada'] <= limite_superior)].copy()

cantidad_eliminados = len(df_limpio) - len(df_sin_outliers)

print(f'Se eliminaron {cantidad_eliminados} filas.')
print(f'No se encontraron outliers')
print('Las columnas observadas son:')
print(' - velocidad_registrada')

# Visualización
plt.boxplot(df_sin_outliers['velocidad_registrada'])
plt.title('Velocidad Registrada sin Outliers')
plt.show()
````

Cálculo de Excesos y Filtrado Final

````python
# Crear la columna exceso_velocidad_real y calcularla. El exceso de velocidad real es la diferencia entre velocidad_registrada y la velocidad_maxima.
# Mostrar las primeras 10 patentes y su exceso de velocidad real.

df_sin_outliers['exceso_velocidad_real'] = df_sin_outliers['velocidad_registrada'] - df_sin_outliers['velocidad_maxima']


print("Primeras 10 filas:")
display(df_sin_outliers.head(10))
````

````python
# Crear la columna exceso_velocidad y calcularla.
#El exceso de velocidad es la diferencia entre velocidad_registrada y la velocidad_maxima más un 5 porciento de la velocidad máxima.
#Mostrar las primeras 10 patentes y su exceso de velocidad.

df_sin_outliers['exceso_velocidad'] = df_sin_outliers['velocidad_registrada'] - (df_sin_outliers['velocidad_maxima'] * 1.05)

print("--- 10 primeras patentes y su exceso de velocidad ---")
print(df_sin_outliers[['patente', 'exceso_velocidad']].head(10))
````

````python
# Eliminar las filas que no cometieron infracciones según el exceso de velocidad.Mostrar cuantas filas se eliminaron con la leyenda
# Se eliminaron X filas.

# pre_filtro: int = len(df_fines)
# df_fines = df_fines[df_fines['exceso_velocidad'] > 0]
# eliminados_sin_infracc: int = pre_filtro - len(df_fines)

pre_filtro: int = len(df_sin_outliers)
df_sin_outliers = df_sin_outliers[df_sin_outliers['exceso_velocidad'] > 0]
eliminados_sin_infracc: int = pre_filtro - len(df_sin_outliers)

print(f"Se eliminaron {eliminados_sin_infracc} filas.")
````

Definir una clase FineAnalyzer

````python
# Cargamos el dataset NUEVO (el limpio)

ruta_limpio = 'urban_flow/data/interim/speeding_fines.csv'
df_limpio = pd.read_csv(ruta_limpio)
````

````python
#Clase que analiza estadísticamente el dataset de infracciones.
class FineAnalyzer:

  #inicializar clase y encapsular DataFrame
  def __init__(self, df: pd.DataFrame):
      self._df: pd.DataFrame = df.copy()

  #Retorna top 5 de patentes con más multas
  def ranking_patentes_multadas(self) -> pd.DataFrame:
    ranking = self._df['patente'].value_counts().head(5).reset_index()
    ranking.columns = ['patente', 'cantidad']
    ranking.index = ranking.index + 1
    return ranking

  #Retorna top 5 horarios con más multas
  def ranking_horarios_multas(self) -> pd.DataFrame:
    ranking = self._df['hora'].value_counts().head(5).reset_index()
    ranking.columns = ['hora', 'cantidad']
    ranking.index = ranking.index + 1
    return ranking

  #Retorna media de exceso de velocidad reglamentario
  def promedio_exceso_velocidad(self) -> float:
    return float(self._df['exceso_velocidad'].mean())

  #Retorna media del exceso de velocidad real
  def promedio_exceso_real_velocidad(self) -> float:
    return float(self._df['exceso_velocidad_real'].mean())

  #Retorna cant de multas agrupadas por ubicación (orden alfabético)
  def multas_por_ubicacion(self) -> pd.DataFrame:
    conteo = self._df['ubicacion'].value_counts().reset_index()
    conteo.columns = ['ubicacion', 'cantidad']
    # Ordenamos alfabéticamente por ubicación
    conteo_ordenado = conteo.sort_values(by='ubicacion')
    # Reseteamos el índice para que quede prolijo al mostrarlo
    return conteo_ordenado.reset_index(drop=True)
````

### Instanciación del objeto FineAnalyzer

````python
# Creamos el objeto pasando nuestro DataFrame limpio
analizador = FineAnalyzer(df_limpio)
print("Objeto analizador creado con éxito.")
````

### Top 5 Patentes más multadas


````python
#Llamamos al top
top_patentes = analizador.ranking_patentes_multadas()
display(top_patentes)
````

### Top 5 horarios con + multas

````python
#Llamamos al ranking de la class
top_horarios = analizador.ranking_horarios_multas()
display(top_horarios)
````

### Media excesos de velocidad

````python
promedio_reglamentario = analizador.promedio_exceso_velocidad()
promedio_real = analizador.promedio_exceso_real_velocidad()

print(f"Exceso promedio (reglamentario): {promedio_reglamentario:.2f} km/h")
print(f"Exceso promedio (real): {promedio_real:.2f} km/h")
````

### Multas por ubicación

````python
#Llamamos al analizador
ubicaciones = analizador.multas_por_ubicacion()
display(ubicaciones)
````
